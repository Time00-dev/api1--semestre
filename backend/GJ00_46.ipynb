```python
import time

def baixar_com_retry(url, SESSION, TIMEOUT):  #Leitura geral de arquivos
  primeria_tentativa = True
  headers_req = {
      "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
      "Accept": "application/json, text/plain, */*",
      "Accept-Language": "pt-BR,pt;q=0.9,en-US;q=0.8,en;q=0.7",
      "Referer": "https://olinda.bcb.gov.br/",
  }
  for i in range(1, MAX_GET_RETRIES + 1): #Faz get com limite de tentativas
    try:
      resposta = SESSION.get(url, headers=headers_req, timeout=TIMEOUT) #Tenta em um tempo curto
      resposta.raise_for_status()
      resposta.encoding = "utf-8-sig"
      return resposta
    except Exception as e:
      if i < MAX_GET_RETRIES:
        time.sleep(0.5 * i)
      else:
        print(f"Todas as {MAX_GET_RETRIES} tentativas falharam: {e}")
        return None
```

```python
def ler_json(url, SESSION, timeout = 5, chave = None, funcao_tratamento = None, filtro_colunas = None):
  resposta = baixar_com_retry(url, SESSION, timeout)
  if resposta is None:
    return

  json = resposta.json()

  if(chave and chave in json):
    df = pd.DataFrame(json[chave])
  else:
    df = pd.DataFrame(json)

  if filtro_colunas:
    df = df[filtro_colunas]
  if funcao_tratamento:
    return funcao_tratamento(df)

  return df
```

#**Função para enviar dados**

```python
# [ALTERADO] (GJ00-60) Envio dos resultados para a API Flask, que grava no banco (POST /api/carga).
# Antes: uma requisição por linha, com os cabeçalhos do BCB. Agora: uma requisição por mês, com um token.
# O formato é o mesmo que a API devolve ao site (frontend/dados/mock_sub_regioes.json).
CAMPOS_CHAVE = ["ano_mes", "regiao", "uf", "sub_regiao"]
CAMPOS_EIXO_I = ["inadimplenciaReal", "fragilidadeRenda", "agingDivida", "vulnerabilidadeSocial"]
CAMPOS_EIXO_II = ["maturidadePix", "crescimentoPopulacional", "totalHabitantes", "bonusDemografico"]
CAMPOS_SCORE = ["score", "faixa", "ranking"]


def _valor_json(valor):
    """Vazio (NaN) vira null e números do numpy viram números comuns: o JSON não aceita NaN."""
    if valor is None or (not isinstance(valor, str) and pd.isna(valor)):
        return None
    if isinstance(valor, np.integer):
        return int(valor)
    if isinstance(valor, (np.floating, float)):
        return round(float(valor), 2)
    return valor


def montar_lotes(df_score):
    """Um lote por mês no formato do contrato: {"eixo_i": [...], "eixo_ii": [...]}, com score, faixa e ranking."""
    lotes = {}
    for mes, grupo in df_score.groupby("ano_mes", sort=True):
        # a API recusa o mês inteiro se uma linha tiver chave inválida; essas linhas ficam fora e o aviso mostra quantas
        chave_ok = (grupo["uf"].astype(str).str.fullmatch(r"[A-Z]{2}")
                    & grupo["sub_regiao"].astype(str).str.fullmatch(r"\d{2}"))
        if not chave_ok.all():
            print(f"{mes}: {int((~chave_ok).sum())} linha(s) sem UF ou sub-região válida ficaram fora do envio")
        linhas = grupo[chave_ok].to_dict("records")
        lotes[str(mes)] = {
            eixo: [{c: _valor_json(linha[c]) for c in CAMPOS_CHAVE + campos + CAMPOS_SCORE} for linha in linhas]
            for eixo, campos in (("eixo_i", CAMPOS_EIXO_I), ("eixo_ii", CAMPOS_EIXO_II))
        }
    return lotes


def enviar_resultados(df_score, url_api, token):
    """Manda mês a mês para o Flask (ligado no PC, com o túnel). Para no primeiro erro, sem esconder o motivo."""
    if not url_api or not token:
        raise ValueError("Preencha URL_API_LOCAL com o link do túnel e crie o secret CARGA_TOKEN no Colab")
    total = 0
    for mes, lote in montar_lotes(df_score).items():
        resposta = requests.post(url_api.rstrip("/") + "/api/carga", json=lote,
                                 headers={"X-Token-Carga": token}, timeout=60)
        if resposta.status_code != 201:
            raise RuntimeError(f"Mês {mes}: erro {resposta.status_code} - {resposta.text[:300]}")
        gravadas = resposta.json()["linhas_gravadas"]
        total += gravadas
        print(f"{mes}: {gravadas} linhas gravadas")
    print(f"Total: {total} linhas gravadas no banco")
    return total
```

```python
#Funções gerais de tratamento de dados

def limpeza_string(df, colunas=None):
    if colunas is None:
        colunas = df.select_dtypes(include="object").columns.tolist()
    else:
        colunas = [col for col in colunas if col in df.columns and df[col].dtype == "object"]

    if not colunas:
        return df

    for col in colunas:
        df[col] = df[col].str.strip()
    return df

def converter_numerico(df, *colunas):
    for col in colunas:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

def limpar_nulos(df):
    numericas = df.select_dtypes(include="number").columns
    df[numericas] = df[numericas].fillna(df[numericas].median())

    categoricas = df.select_dtypes(include="object").columns
    df[categoricas] = df[categoricas].fillna("NAO_INFORMADO")
    return df

def formatar_data(df, colunas, formato="%Y-%m-%d"):
    if not colunas: return df
    if isinstance(colunas, str):
        colunas = [colunas]
    for col in colunas:
        df[col] = pd.to_datetime(df[col], format=formato).dt.strftime("%Y%m")
    return df

def formatar_numero(df, colunas):
    if isinstance(colunas, str):
        colunas = [colunas]
    for col in colunas:
        df[col] = pd.to_numeric(df[col].str.replace(",", "."), errors="coerce")
    return df

# [REMOVIDO] def converter_nivel_salario(df, coluna): ... (dependia do NIVEIS_SALARIO; não é mais chamada)
# def converter_nivel_salario(df, coluna):
#   df[coluna] = df[coluna].map(NIVEIS_SALARIO)
#   return df

def converter_uf(df, coluna):
  estados_brasil = {
    "Acre": "AC", "Alagoas": "AL", "Amapá": "AP", "Amazonas": "AM",
    "Bahia": "BA", "Ceará": "CE", "Distrito Federal": "DF", "Espírito Santo": "ES",
    "Goiás": "GO", "Maranhão": "MA", "Mato Grosso": "MT", "Mato Grosso do Sul": "MS",
    "Minas Gerais": "MG", "Pará": "PA", "Paraíba": "PB", "Paraná": "PR",
    "Pernambuco": "PE", "Piauí": "PI", "Rio de Janeiro": "RJ", "Rio Grande do Norte": "RN",
    "Rio Grande do Sul": "RS", "Rondônia": "RO", "Roraima": "RR", "Santa Catarina": "SC",
    "São Paulo": "SP", "Sergipe": "SE", "Tocantins": "TO",
    11: "RO", 12: "AC", 13: "AM", 14: "RR",
    15: "PA", 16: "AP", 17: "TO", 21: "MA",
    22: "PI", 23: "CE", 24: "RN", 25: "PB",
    26: "PE", 27: "AL", 28: "SE", 29: "BA",
    31: "MG", 32: "ES", 33: "RJ", 35: "SP",
    41: "PR", 42: "SC", 43: "RS", 50: "MS",
    51: "MT", 52: "GO", 53: "DF"
  }

  df = df.rename(columns = {
      coluna: "uf"
  })
  df["uf"] = df["uf"].map(estados_brasil)

  return df

def converter_uf_para_macro_regiao(df, coluna):
  mapeamento_regioes = {
      'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte', 'RR': 'Norte', 'TO': 'Norte',
      'AL': 'Nordeste', 'BA': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
      'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste',
      'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
      'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
      'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
  }

  df['regiao'] = df[coluna].map(mapeamento_regioes)

  return df

def converter_porcentagem(df, coluna):
  df[coluna] = pd.to_numeric(df[coluna], errors='coerce') / 100
  return df
```

```python
def validar_consistencia_saldos(df):
    """Função para checar se a soma dos saldos por modalidade é igual à carteira ativa total."""
    # 1. Filtra registros de carteira ativa
    df_ativa = df[df["metrica"] == "carteira_ativa"]

    # 2. Agrupa por modalidade e calcula a soma total
    soma_modalidades = (
        df_ativa.groupby(["ano_mes", "uf", "modalidade"])["valor"]
        .sum()
        .groupby(["ano_mes", "uf"])
        .sum()
        .reset_index(name="soma_modalidades")
    )

    # 3. Obtém o total da carteira ativa informado para o período/UF
    total_carteira = (
        df_ativa.groupby(["ano_mes", "uf"])["valor"]
        .sum()
        .reset_index(name="total_carteira_ativa")
    )

    # 4. Cruza e valida a diferença
    checagem = pd.merge(soma_modalidades, total_carteira, on=["ano_mes", "uf"])
    checagem["diferenca"] = (
        checagem["soma_modalidades"] - checagem["total_carteira_ativa"]
    ).round(2)

    inconsistencias = checagem[checagem["diferenca"] != 0]

    if not inconsistencias.empty:
        print(
            f"[VALIDAÇÃO] Encontradas {len(inconsistencias)} inconsistências entre modalidades e carteira ativa!"
        )
    else:
        print(
            "[VALIDAÇÃO] Saldos entre modalidades e carteira ativa estão 100% consistentes!"
        )

def padronizar_df_scr(df_scr):
    df_scr = df_scr.rename(columns={"data_base": "ano_mes"})                      # [IGUAL]

    df_scr["tipo"] = df_scr["cliente"].map({"PF": "pf"}).fillna("pj")             # [IGUAL]

    df_scr = df_scr.melt(
        id_vars=["ano_mes", "regiao", "uf", "sub_regiao", "tipo", "classe", "modalidade"],   # [ALTERADO] + "sub_regiao"
        value_vars=["carteira_vencida", "carteira_ativa"],                        # [IGUAL]
        var_name="metrica",
        value_name="valor",
    )
    df_scr["origem"] = "scr"                                                      # [IGUAL]

     # [REMOVIDO] validar_consistencia_saldos(df_scr): comparava a soma do dado com ela mesma e rodava por chunk (mensagem repetida)

    for col in ["tipo", "classe", "origem", "metrica", "uf", "sub_regiao", "modalidade"]:   # [ALTERADO] + "sub_regiao"
        if col in df_scr.columns:
            df_scr[col] = df_scr[col].astype("category")

    df_scr = (
        df_scr.groupby(
            [
                "ano_mes",
                "regiao",
                "uf",
                "sub_regiao",                                                     # [NOVO]
                "tipo",
                "classe",
                "modalidade",
                "metrica",
                "origem",
            ],
            observed=True,
        )["valor"]
        .sum()
        .reset_index()
    )

    return df_scr[[
        "ano_mes",
        "regiao",
        "uf",
        "sub_regiao",                                                             # [NOVO]
        "tipo",
        "classe",
        "metrica",
        "origem",
        "valor",
    ]]
```

```python
def validar_scr_contra_origem(caminho_origem="/content/dados/scr_data.parquet",
                              caminho_final="/content/dados/scr_pix.parquet"):   # [NOVO]
    origem = pd.read_parquet(caminho_origem)
    origem["data_base"] = origem["data_base"].astype(str)
    esperado = origem.groupby("data_base")[["carteira_ativa", "carteira_vencida"]].sum()

    final = pd.read_parquet(caminho_final)
    final = final[final["origem"] == "scr"]
    obtido = final.groupby(["ano_mes", "metrica"], observed=True)["valor"].sum().unstack()

    dif = (esperado - obtido[["carteira_ativa", "carteira_vencida"]]).abs()
    erros = dif[(dif > 1).any(axis=1)]

    if erros.empty:
        print(f"[VALIDAÇÃO] SCR: soma da origem = soma do scr_pix em {len(esperado)} meses.")
    else:
        print(f"[VALIDAÇÃO] SCR: divergência em {len(erros)} meses (diferença em R$):")
        print(erros)
```

```python
def padronizar_df_pix(df_pix):
    df_pix = df_pix.rename(columns={"AnoMes": "ano_mes"})

    df_pix = df_pix.melt(
        id_vars=["ano_mes", "regiao", "uf"],
        value_vars=["VL_PagadorPF", "QT_PagadorPF", "VL_PagadorPJ", "QT_PagadorPJ", "QT_PES_PagadorPF", "QT_PES_PagadorPJ"],
        var_name="metrica_raw",
        value_name="valor"
    )

    df_pix["tipo"] = df_pix["metrica_raw"].str.endswith("PF").map({True: "pf", False: "pj"})

    df_pix["metrica"] = np.select(
        [df_pix["metrica_raw"].str.contains("VL"),
         df_pix["metrica_raw"].str.contains("QT_PES"),
         df_pix["metrica_raw"].str.contains("QT")],
        ["vl_pagador", "qt_pes_pagador", "qt_pagador"],
        default="outros"
    )

    df_pix["origem"] = "pix"

    df_pix = df_pix.groupby(
        ["ano_mes", "regiao", "uf", "tipo", "metrica", "origem"],
        observed=True
    )["valor"].sum().reset_index()

    df_pix["classe"] = pd.NA  # adiciona depois do groupby

    for col in ["tipo", "origem", "metrica", "uf", "regiao"]:
        df_pix[col] = df_pix[col].astype("category")

    return df_pix[["ano_mes", "regiao", "uf", "tipo", "classe", "metrica", "origem", "valor"]]
```

```python
def construir_df_scr_pix():
  if not Path('/content/dados/pix_transacoes.parquet').exists() or not Path('/content/dados/scr_data.parquet').exists():
      raise FileNotFoundError("Arquivos pix_transacoes.parquet ou scr_data.parquet não foram gerados pela API do BCB. Verifique a conexão com a API do Banco Central na etapa anterior.")

  chunks_pix = []
  chunks_scr = []

  #Padroniza o parquet pix por chunk
  tabela_pix = pq.ParquetFile('/content/dados/pix_transacoes.parquet')

  for batch in tabela_pix.iter_batches(batch_size=100000):
    chunks_pix.append(padronizar_df_pix(batch.to_pandas()))
  df_pix = pd.concat(chunks_pix, ignore_index=True)

  df_pix = df_pix.groupby(                                                   # [IGUAL] groupby do Pix (sem sub_regiao)
      ["ano_mes", "regiao", "uf", "tipo", "metrica", "origem"],
      observed=True
  )["valor"].sum().reset_index()
  df_pix["classe"] = pd.NA                                                   # [IGUAL]
  df_pix["sub_regiao"] = pd.NA                                               # [NOVO]

  #Padroniza o parquet scr por chunk
  tabela_scr = pq.ParquetFile('/content/dados/scr_data.parquet')

  for batch in tabela_scr.iter_batches(batch_size=100000):
    chunks_scr.append(padronizar_df_scr(batch.to_pandas()))
  df_scr = pd.concat(chunks_scr, ignore_index=True)

  df_scr = df_scr.groupby(
      ["ano_mes", "regiao", "uf", "sub_regiao", "tipo", "classe", "metrica", "origem"],   # [ALTERADO] + "sub_regiao"
      observed=True
  )["valor"].sum().reset_index()

  df_pix["ano_mes"] = df_pix["ano_mes"].astype(str)                          # [IGUAL]
  df_scr["ano_mes"] = df_scr["ano_mes"].astype(str)                          # [IGUAL]

  #une os dois parquets em um individual
  df_scr_pix = pd.concat([df_scr, df_pix], ignore_index=True)                # [IGUAL]

  salvar_parquet(df_scr_pix, "scr_pix")                                      # [IGUAL]
```

```python
def calculo_idade():
  #---populacao por idade
  df_idade = pd.read_parquet('/content/dados/populacao_por_idade.parquet')

  df_idade["idade_ponderada"] = df_idade["idade"] * df_idade["valor"]

  df_idade = (
      df_idade.groupby(["ano", "regiao", "uf"])
      .agg(
          soma_idade=("idade_ponderada", "sum"),
          soma_pop=("valor", "sum")
      )
      .reset_index()
  )

  df_idade["idade_media"] = df_idade["soma_idade"] / df_idade["soma_pop"]
  # [ALTERADO] Removido o cálculo de diferença (idade_max - idade_media);
  # Agora atribue a idade media direto ao indicador, mantendo o nome da coluna por compatibilidade
  df_idade["bonusDemografico"] = df_idade["idade_media"]

  return df_idade[["ano", "regiao", "uf", "bonusDemografico"]]
```

```python
def construir_df_ibge():
  #---escolarizacao
  df_escolarizacao = extract_data('/content/dados/taxa_escolarizacao.parquet')

  df_escolarizacao = df_escolarizacao.groupby(  #realiza a média de valores com mesmo ano, regiao e uf
      ["ano", "regiao", "uf"],
      observed=True)["valor"].mean().reset_index()
  df_escolarizacao = df_escolarizacao.rename(columns={"valor": "taxa_escolarizacao"})

  #---censo demografico
  df_censo_demografico = extract_data('/content/dados/censo_demografico.parquet')

  #separa colunas por metrica
  df_censo_demografico = df_censo_demografico.pivot_table(
      index=["ano", "regiao", "uf"],
      columns="D2N",
      values="valor").reset_index()

  df_censo_demografico = df_censo_demografico.rename(columns={
      "População residente": "populacao_residente",
      "Taxa de crescimento geométrico": "taxa_crescimento",
      "Variação absoluta da população residente 2010 compatibilizada": "variacao_populacao"
  })

  df_idade = calculo_idade()

  #arruma porcentagem
  df_censo_demografico["taxa_crescimento"] = df_censo_demografico["taxa_crescimento"]/100

  df_censo_demografico.columns.name = None

  # [NOVO] Censo e idade só existem em anos de censo (2010, 2022), então não casam com os anos da escolarização.
  # Fica com o dado mais recente de cada UF (.last()) e tira a coluna "ano", para valer para todos os anos
  df_censo_demografico = (df_censo_demografico.sort_values("ano")
      .groupby(["regiao", "uf"]).last().reset_index().drop(columns=["ano"]))
  df_idade = (df_idade.sort_values("ano")
      .groupby(["regiao", "uf"]).last().reset_index().drop(columns=["ano"]))

  # [ALTERADO] Antes cruzava por ["ano", "regiao", "uf"]. Agora cruza só por ["regiao", "uf"]
  df_ibge = df_escolarizacao.merge(
      df_censo_demografico,
      on=["regiao", "uf"],
      how="left").merge(
      df_idade,
      on=["regiao", "uf"],
      how="left")

  # [ALTERADO] Ordena por UF, região e ano, agrupando por UF antes de aplicar fill/bfill;
  # Isso garante que o preenchimento de nulos use apenas dados do mesmo estado, sem vazar para outras UFs
  df_ibge = df_ibge.sort_values(["uf", "regiao", "ano"]).reset_index(drop=True)
  colunas_preencher = ["taxa_escolarizacao", "populacao_residente", "taxa_crescimento", "variacao_populacao", "bonusDemografico"]
  df_ibge[colunas_preencher] = df_ibge.groupby("uf")[colunas_preencher].transform(lambda x: x.ffill().bfill())

  salvar_parquet(df_ibge, "ibge")
```

```python
def normalizacao(df, coluna):
  df[coluna + "_max"] = df.groupby("ano_mes")[coluna].transform("max")
  df[coluna + "_min"] = df.groupby("ano_mes")[coluna].transform("min")
  df[coluna] = (1 + (df[coluna] - df[coluna + "_min"]) / (df[coluna + "_max"] - df[coluna + "_min"]) * 4).round(2)
  df = df.drop(columns=[coluna + "_max", coluna + "_min"])

  return df
```

```python
filtro = ["ano_mes", "regiao", "uf"]


def agrupar_metrica(df, metrica, filtro_extra=None):
    mask = df["metrica"] == metrica
    if filtro_extra is not None:
        mask &= filtro_extra
    return df[mask].groupby(filtro)["valor"].sum()
```

```python
def calcular_crescimento_vencida(df_scr_pix_base):
    base = df_scr_pix_base[df_scr_pix_base["metrica"] == "carteira_vencida"]
    atual = (
        base.groupby(["ano_mes", "regiao", "uf"], observed=True)["valor"]
        .sum()
        .reset_index()
    )
    atual = atual.rename(columns={"valor": "vencida"})

    atual["ano_mes"] = atual["ano_mes"].astype(str)
    atual["uf"] = atual["uf"].astype(str)
    atual["regiao"] = atual["regiao"].astype(str)
    atual["data"] = pd.to_datetime(atual["ano_mes"], format="%Y%m")

    passado = atual[["uf", "data", "vencida"]].rename(
        columns={"vencida": "vencida_anterior"}
    )
    passado["data"] = passado["data"] + pd.DateOffset(
        months=PERIODOS_CRESCIMENTO
    )

    atual = atual.merge(passado, on=["uf", "data"], how="left")

    atual["agingDivida"] = np.where(
        atual["vencida_anterior"] > 0,
        (atual["vencida"] / atual["vencida_anterior"]) - 1,
        0.0,
    )
    return atual[["ano_mes", "regiao", "uf", "agingDivida"]]


def calcular_eixo_i(df_scr_pix_base, df_ibge_base):
    df_scr_pix = df_scr_pix_base.copy()
    df_ibge = df_ibge_base.copy()

    df_crescimento = calcular_crescimento_vencida(df_scr_pix_base)

    df_scr_pix["ano"] = df_scr_pix["ano_mes"].astype(str).str[:4]

    df_calculos = df_scr_pix.merge(
        df_ibge[["ano", "regiao", "uf", "taxa_escolarizacao"]],
        on=["ano", "regiao", "uf"],
        how="left",
    ).drop(columns=["ano"])
        # [NOVO] Ordena por UF e mês: o ffill/bfill abaixo preenche conforme a ordem das linhas.
    # Sem isso, 2021 e 2022 receberiam a escolarização de 2025 em vez da de 2023
    df_calculos = df_calculos.sort_values(["uf", "ano_mes"])

    df_calculos["taxa_escolarizacao"] = df_calculos.groupby("uf")[
        "taxa_escolarizacao"
    ].transform(lambda x: x.ffill().bfill())

    # --- CORREÇÃO 1: Removida a duplicação do agrupar_metrica da carteira_ativa_classes_de ---   # [IGUAL]
    df_metricas = pd.concat(
        [
            agrupar_metrica(df_calculos, "carteira_vencida").rename(
                "carteira_vencida"
            ),                                                                                    # [IGUAL]
            agrupar_metrica(df_calculos, "carteira_ativa").rename(
                "carteira_ativa"
            ),                                                                                    # [IGUAL]

            # [NOVO] DECISÃO (21/10/2026): a API do SCR traz RISCO de crédito (AA-C, D-H, N/D),
            # e não mais a classe de renda (A a E). Por isso "D-H" (risco alto) passa a valer
            # no lugar de "D" e "E". O nome carteira_ativa_classes_de e o indicador
            # fragilidadeRenda foram mantidos para não quebrar o resto do código.
            agrupar_metrica(
                df_calculos, "carteira_ativa", df_calculos["classe"].isin(["D-H"])   # [ALTERADO] antes: .isin(["D", "E"])
            ).rename("carteira_ativa_classes_de"),
        ],
        axis=1,
    ).reset_index()                                                                               # [IGUAL]
    df_calculos = (
        df_calculos[["ano_mes", "regiao", "uf", "taxa_escolarizacao"]]
        .drop_duplicates()
        .merge(df_metricas, on=filtro, how="left")
    )
      # [NOVO] Remove meses que só existem no Pix (202011, 202012, 202609): sem SCR não há carteira
    # e esses meses apareceriam com zeros falsos
    df_calculos = df_calculos.dropna(subset=["carteira_ativa"])

    df_calculos = df_calculos.merge(df_crescimento, on=filtro, how="left")
    df_calculos["agingDivida"] = df_calculos["agingDivida"].fillna(0.0)

    # --- Divisões com tratamento de divisão por zero ---
    df_calculos["inadimplenciaReal"] = np.where(
        df_calculos["carteira_ativa"] > 0,
        df_calculos["carteira_vencida"] / df_calculos["carteira_ativa"],
        0.0,
    )

    # [NOVO] DECISÃO (21/10/2026): a partir de 202501 a API só envia o risco "N/D" (não há "D-H").
    # Nesses meses todas as UFs ficam com fração 0, a normalizacao gera NaN (0/0) e o fillna(0)
    # final transforma em 0. A escala válida é de 1 a 5, então 0 significa "SEM DADO", e não "risco nulo".
    df_calculos["fragilidadeRenda"] = np.where(                                  # [IGUAL]
        df_calculos["carteira_ativa"] > 0,
        df_calculos["carteira_ativa_classes_de"] / df_calculos["carteira_ativa"],
        0.0,
    )

    # --- CORREÇÃO 2: Sintaxe de vulnerabilidadeSocial corrigida ---
    df_calculos["vulnerabilidadeSocial"] = 1 - df_calculos["taxa_escolarizacao"]

    # --- Normalização ---
    df_calculos = normalizacao(df_calculos, "inadimplenciaReal")
    df_calculos = normalizacao(df_calculos, "fragilidadeRenda")
    df_calculos = normalizacao(df_calculos, "agingDivida")
    df_calculos = normalizacao(df_calculos, "vulnerabilidadeSocial")

    # --- Limpeza final ---
    df_calculos = df_calculos.replace([np.inf, -np.inf], np.nan).fillna(0)

    return df_calculos[[
        "ano_mes",
        "regiao",
        "uf",
        "inadimplenciaReal",
        "fragilidadeRenda",
        "agingDivida",
        "vulnerabilidadeSocial",
    ]]
```

```python
def calcular_eixo_ii(df_scr_pix_base, df_ibge_base):
    df_scr_pix = df_scr_pix_base.copy()
    df_ibge = df_ibge_base.copy()

    # --- Extrai ano e merge com IBGE ---
    df_scr_pix["ano"] = df_scr_pix["ano_mes"].astype(str).str[:4]

    df_calculos = df_scr_pix.merge(
        df_ibge[["ano", "regiao", "uf", "populacao_residente", "taxa_crescimento", "variacao_populacao", "bonusDemografico"]],
        on=["ano", "regiao", "uf"],
        how="left"
    ).drop(columns=["ano"])

    # [NOVO] Ordena por UF e mês antes de preencher nulos para manter a cronologia correta dentro de cada estado
    df_calculos = df_calculos.sort_values(["uf", "ano_mes"])

    def eliminar_duplicata(metrica):
        df_calculos[metrica] = (
            df_calculos.groupby("uf")[metrica]
            .transform(lambda x: x.ffill().bfill())
        )

    eliminar_duplicata("populacao_residente")
    eliminar_duplicata("taxa_crescimento")
    eliminar_duplicata("variacao_populacao")
    eliminar_duplicata("bonusDemografico")

    # --- Agrupamento das métricas Pix ---
    df_metricas = pd.concat([
        agrupar_metrica(df_calculos, "vl_pagador").rename("vl_pagador"),
        agrupar_metrica(df_calculos, "qt_pes_pagador").rename("qt_pes_pagador"),
        agrupar_metrica(df_calculos, "qt_pagador").rename("qt_pagador"),
    ], axis=1).reset_index()

    df_calculos = (
        df_calculos[["ano_mes", "regiao", "uf", "populacao_residente", "taxa_crescimento", "variacao_populacao", "bonusDemografico"]]
        .drop_duplicates()
        .merge(df_metricas, on=filtro, how="left")
    )

    # [ALTERADO] Salva as duas métricas brutas em colunas para normalizá-las separadamente
    # 1. Maturidade Uso PIX = qt_pagador / qt_pes_pagador
    df_calculos["maturidade_uso"] = np.where(
        df_calculos["qt_pes_pagador"] > 0,
        df_calculos["qt_pagador"] / df_calculos["qt_pes_pagador"],
        0.0
    )

    # 2. Maturidade Financeira PIX = vl_pagador / qt_pagador
    df_calculos["maturidade_financeira"] = np.where(
        df_calculos["qt_pagador"] > 0,
        df_calculos["vl_pagador"] / df_calculos["qt_pagador"],
        0.0
    )

    # [NOVO] Normaliza cada métrica na escala de 1 a 5 antes de aplicar os pesos,
    # evitando que o ticket médio em reais domine a frequência de uso
    df_calculos = normalizacao(df_calculos, "maturidade_uso")
    df_calculos = normalizacao(df_calculos, "maturidade_financeira")

    # 3. Consolidado Maturidade PIX = (Uso * 0.6) + (Financeira * 0.4)
    df_calculos["maturidadePix"] = (df_calculos["maturidade_uso"] * 0.6) + (df_calculos["maturidade_financeira"] * 0.4)

    # Demais indicadores do Eixo II
    df_calculos["crescimentoPopulacional"] = df_calculos["taxa_crescimento"]
    df_calculos["totalHabitantes"] = df_calculos["populacao_residente"]

    # --- Normalização ---
    df_calculos = normalizacao(df_calculos, "crescimentoPopulacional")
    df_calculos = normalizacao(df_calculos, "totalHabitantes")
    df_calculos = normalizacao(df_calculos, "bonusDemografico")

    # --- Limpeza final de segurança ---
    df_calculos = df_calculos.replace([np.inf, -np.inf], np.nan).fillna(0)

    return df_calculos[["ano_mes", "regiao", "uf", "maturidadePix", "crescimentoPopulacional", "totalHabitantes", "bonusDemografico"]]
```

# **Score por Sub-região – Oportunidade de Crédito**
---
Score de 0 a 100 por UF + sub-região e mês, com faixa e posição no ranking do mês.

```python
# [NOVO] Célula inteira nova: função do score por sub-região.
# SCORE POR SUB-REGIÃO — oportunidade de crédito, de 0 a 100

# FÓRMULA  [ALTERADO] (GJ00-60) pesos decididos com o professor em 09/11: mais peso para a capacidade de pagamento
#   O (oportunidade) = (média dos indicadores do Eixo II - 1) / 4      -> de 0 a 1, quanto maior, melhor
#   C (segurança)    = 1 - (média dos indicadores do Eixo I - 1) / 4   -> de 0 a 1, quanto maior, menor o risco
#   score            = 100 x O^0,4 x C^0,6                             -> de 0 a 100
#   Exemplo: média Eixo II = 3,4 e média Eixo I = 2,6  ->  O = 0,6, C = 0,6  ->  score = 60
#   Mercado grande com risco alto (O = 0,9, C = 0,1) -> 24,08 (a média geométrica simples daria 30)
#
# INDICADORES (notas de 1 a 5; 0 nas tabelas dos eixos = SEM DADO e fica fora da média):
#   Eixo I  (risco, maior = pior):        inadimplenciaReal, fragilidadeRenda, agingDivida, vulnerabilidadeSocial
#   Eixo II (oportunidade, maior = melhor): maturidadePix, crescimentoPopulacional, totalHabitantes, bonusDemografico
#
# NÍVEL E FONTE:  [ALTERADO] (GJ00-60) chave UF + sub-região, a mesma da tabela do banco, da API e do site
#   Eixo I  -> calculado POR UF + SUB-REGIÃO a partir do SCR (df_scr_pix), mesma metodologia do calcular_eixo_i,
#              somando as carteiras antes de dividir. Escolarização (IBGE) é a da própria UF.
#   Eixo II -> existe só POR UF (Pix e IBGE): cada sub-região herda as notas da própria UF.
#   Sub-região = UF + 2 primeiros dígitos do CEP. O mesmo código pode existir em UFs diferentes
#              (o 69 existe no AC, AM e RR) e cada par é uma sub-região separada.

import numpy as np
import pandas as pd

INDICADORES_EIXO_I = ["inadimplenciaReal", "fragilidadeRenda", "agingDivida", "vulnerabilidadeSocial"]
INDICADORES_EIXO_II = ["maturidadePix", "crescimentoPopulacional", "totalHabitantes", "bonusDemografico"]
MIN_INDICADORES_POR_EIXO = 1      # mínimo de indicadores com dado em cada eixo para calcular o score

PESO_OPORTUNIDADE = 0.4         # [NOVO] (GJ00-60) expoente de O na fórmula
PESO_CAPACIDADE = 0.6           # [NOVO] (GJ00-60) expoente de C na fórmula

# [ALTERADO] (GJ00-60) Faixas do relatório da Sprint 3 (convenção do time; podem ser ajustadas com o cliente).
# (limite inferior, limite superior, nome). Inferior incluído, superior não (a última faixa inclui o 100).
FAIXAS_SCORE = [
    (0, 30, "Não recomendado"),
    (30, 50, "Baixa atratividade"),
    (50, 70, "Oportunidade com cautela"),
    (70, 100, "Oportunidade prioritária"),
]

SUB_REGIAO_INVALIDA = ["nan", "None", "", "0"]   # o pipeline troca NaN por 0 no final


def score_de_medias(media_eixo_i, media_eixo_ii):
    """Fórmula decidida com o professor. Recebe as médias (notas de 1 a 5) e devolve (O, C, score).
    Aceita número ou Series."""
    o = ((media_eixo_ii - 1) / 4).clip(0, 1) if hasattr(media_eixo_ii, "clip") else min(max((media_eixo_ii - 1) / 4, 0), 1)
    c = (1 - (media_eixo_i - 1) / 4).clip(0, 1) if hasattr(media_eixo_i, "clip") else min(max(1 - (media_eixo_i - 1) / 4, 0), 1)
    return o, c, np.round(100 * o ** PESO_OPORTUNIDADE * c ** PESO_CAPACIDADE, 2)  # [ALTERADO] (GJ00-60)


def _escolarizacao_uf_mes(df_meses_uf, df_ibge):
    """taxa_escolarizacao por (ano_mes, uf), com a mesma regra do calcular_eixo_i (ffill/bfill por UF)."""
    base = df_meses_uf[["ano_mes", "uf"]].drop_duplicates().copy()
    base["ano"] = base["ano_mes"].str[:4]
    ibge = df_ibge[["ano", "uf", "taxa_escolarizacao"]].copy()
    ibge["ano"] = ibge["ano"].astype(str)
    ibge["uf"] = ibge["uf"].astype(str)
    ibge["taxa_escolarizacao"] = ibge["taxa_escolarizacao"].where(ibge["taxa_escolarizacao"] > 0)  # 0 = sem dado
    ibge = ibge.groupby(["ano", "uf"], as_index=False)["taxa_escolarizacao"].mean()
    out = base.merge(ibge, on=["ano", "uf"], how="left").sort_values(["uf", "ano_mes"])
    out["taxa_escolarizacao"] = out.groupby("uf")["taxa_escolarizacao"].transform(lambda x: x.ffill().bfill())
    return out[["ano_mes", "uf", "taxa_escolarizacao"]]


def calcular_eixo_i_por_nivel(df_scr_pix_base, df_ibge_base, nivel=("sub_regiao",)):
    """Eixo I (notas de 1 a 5; 0 = sem dado) mês a mês no nível pedido.
    nivel=("sub_regiao",) é o usado no score; ("regiao", "uf") reproduz o calcular_eixo_i."""
    nivel = list(nivel)
    chave = ["ano_mes"] + nivel

    scr = df_scr_pix_base[
        (df_scr_pix_base["origem"].astype(str) == "scr")
        & (df_scr_pix_base["metrica"].astype(str).isin(["carteira_ativa", "carteira_vencida"]))
    ].copy()
    for c in ["ano_mes", "regiao", "uf", "sub_regiao", "classe", "metrica"]:
        scr[c] = scr[c].astype(str)
    scr["valor"] = pd.to_numeric(scr["valor"], errors="coerce").fillna(0.0)
    if "sub_regiao" in nivel:
        scr = scr[~scr["sub_regiao"].isin(SUB_REGIAO_INVALIDA)]

    ativa = scr[scr["metrica"] == "carteira_ativa"]
    vencida = scr[scr["metrica"] == "carteira_vencida"]
    ativa_dh = ativa[ativa["classe"] == "D-H"]       # mesma regra do calcular_eixo_i (risco alto)

    df = pd.concat(
        [
            ativa.groupby(chave)["valor"].sum().rename("carteira_ativa"),
            vencida.groupby(chave)["valor"].sum().rename("carteira_vencida"),
            ativa_dh.groupby(chave)["valor"].sum().rename("carteira_ativa_dh"),
        ],
        axis=1,
    ).fillna(0.0).reset_index()

    # agingDivida: crescimento da carteira vencida vs. PERIODOS_CRESCIMENTO meses antes
    df["data"] = pd.to_datetime(df["ano_mes"], format="%Y%m")
    passado = df[nivel + ["data", "carteira_vencida"]].rename(columns={"carteira_vencida": "vencida_anterior"})
    passado["data"] = passado["data"] + pd.DateOffset(months=PERIODOS_CRESCIMENTO)
    df = df.merge(passado, on=nivel + ["data"], how="left")
    # Divisão por zero / sem base de comparação = NaN (sem dado). Com 0.0 a normalização trataria como o menor
    # valor e daria nota 1 (risco mínimo): falta de informação virando risco zero.
    df["agingDivida"] = np.where(df["vencida_anterior"] > 0, df["carteira_vencida"] / df["vencida_anterior"] - 1, np.nan)
    df["inadimplenciaReal"] = np.where(df["carteira_ativa"] > 0, df["carteira_vencida"] / df["carteira_ativa"], np.nan)
    df["fragilidadeRenda"] = np.where(df["carteira_ativa"] > 0, df["carteira_ativa_dh"] / df["carteira_ativa"], np.nan)
    df.loc[df["carteira_ativa"] <= 0, "agingDivida"] = np.nan     # sem carteira ativa no mês: não há o que comparar

    # vulnerabilidadeSocial = 1 - escolarização (IBGE, por UF): média das UFs da unidade ponderada pela carteira ativa
    chave_uf = list(dict.fromkeys(chave + ["uf"]))
    pesos = ativa.groupby(chave_uf)["valor"].sum().rename("peso").reset_index()
    pesos = pesos.merge(_escolarizacao_uf_mes(pesos, df_ibge_base), on=["ano_mes", "uf"], how="left")
    pesos = pesos.dropna(subset=["taxa_escolarizacao"])
    pesos["peso_x_taxa"] = pesos["peso"] * pesos["taxa_escolarizacao"]
    esc = pesos.groupby(chave).agg(peso=("peso", "sum"), peso_x_taxa=("peso_x_taxa", "sum")).reset_index()
    esc["taxa_escolarizacao"] = np.where(esc["peso"] > 0, esc["peso_x_taxa"] / esc["peso"], np.nan)
    df = df.merge(esc[chave + ["taxa_escolarizacao"]], on=chave, how="left")
    df["vulnerabilidadeSocial"] = 1 - df["taxa_escolarizacao"]

    for col in INDICADORES_EIXO_I:                   # nota de 1 a 5 (mesma normalizacao do notebook)
        df = normalizacao(df, col)
    df[INDICADORES_EIXO_I] = df[INDICADORES_EIXO_I].replace([np.inf, -np.inf], np.nan).fillna(0)
    return df[chave + INDICADORES_EIXO_I]


def classificar(score, faixas=None):
    """[NOVO] (GJ00-60) Nome da faixa de cada score. Score vazio = "Sem informação"."""
    faixas = FAIXAS_SCORE if faixas is None else faixas
    faixa = pd.Series("Sem informação", index=score.index)
    for i, (inferior, superior, nome) in enumerate(faixas):
        ultima = i == len(faixas) - 1
        dentro = (score >= inferior) & ((score <= superior) if ultima else (score < superior))
        faixa[dentro] = nome
    return faixa


def calcular_score_sub_regiao(df_scr_pix, df_ibge, df_eixo_ii, faixas=None):
    """Score de 0 a 100 por UF + sub-região e mês, com faixa e posição no ranking do mês.
    Devolve os indicadores dos dois eixos junto com médias, O, C, score, faixa e ranking.
    [ALTERADO] (GJ00-60) chave UF + sub-região; o Eixo II vem da própria UF; a coluna de classificação virou "faixa"."""
    eixo_i = calcular_eixo_i_por_nivel(df_scr_pix, df_ibge, ["uf", "sub_regiao"])
    eixo_ii = df_eixo_ii[["ano_mes", "uf", "regiao"] + INDICADORES_EIXO_II].copy()
    for c in ["ano_mes", "uf", "regiao"]:
        eixo_ii[c] = eixo_ii[c].astype(str)
    df = eixo_i.merge(eixo_ii, on=["ano_mes", "uf"], how="left")

    # Médias só com os indicadores que têm informação (0 = sem dado fica fora; nunca vira risco zero)
    for eixo, colunas in (("eixo_i", INDICADORES_EIXO_I), ("eixo_ii", INDICADORES_EIXO_II)):
        com_dado = df[colunas].where(df[colunas] > 0)
        df[f"n_indicadores_{eixo}"] = com_dado.notna().sum(axis=1)
        df[f"media_{eixo}"] = com_dado.mean(axis=1).where(df[f"n_indicadores_{eixo}"] >= MIN_INDICADORES_POR_EIXO)

    df["O"], df["C"], df["score"] = score_de_medias(df["media_eixo_i"], df["media_eixo_ii"])
    df["faixa"] = classificar(df["score"], faixas)
    df["ranking"] = df.groupby("ano_mes")["score"].rank(ascending=False, method="min").astype("Int64")
    df["total_ranqueado"] = df.groupby("ano_mes")["score"].transform("count")

    colunas = (["ano_mes", "uf", "sub_regiao", "regiao"] + INDICADORES_EIXO_I + INDICADORES_EIXO_II +
               ["n_indicadores_eixo_i", "n_indicadores_eixo_ii", "media_eixo_i", "media_eixo_ii",
                "O", "C", "score", "faixa", "ranking", "total_ranqueado"])
    return df[colunas].sort_values(["ano_mes", "ranking", "uf", "sub_regiao"]).reset_index(drop=True)
```

```python
# [NOVO - GJ00-54] Conta as operações de cada sub-região nos últimos 3 meses

# [NOVO] Lê o SCR que foi salvo (já com a coluna operacoes)
df_scr = pd.read_parquet("/content/dados/scr_data.parquet")

# [NOVO] Pega os 3 últimos meses que existem no dado (não fixa a data, então funciona nos próximos meses)
ultimos_3_meses = sorted(df_scr["data_base"].unique())[-3:]
print("Meses usados:", ultimos_3_meses)

# [NOVO] Fica só com esses 3 meses
df_3m = df_scr[df_scr["data_base"].isin(ultimos_3_meses)]

# [NOVO] Soma as operações de cada sub-região em cada mês
# (o código da sub-região se repete entre estados, por isso agrupa por uf + sub_regiao)
ops_mes = df_3m.groupby(["uf", "sub_regiao", "data_base"])["operacoes"].sum().reset_index()

# [NOVO] Tira a média dos 3 meses. O número de operações é um "retrato" do mês:
# somar os 3 meses contaria as mesmas operações 3 vezes
ops_sub = ops_mes.groupby(["uf", "sub_regiao"])["operacoes"].mean().reset_index()

# [NOVO] Resumo: quantas sub-regiões existem e como as operações se distribuem
print("Sub-regiões:", len(ops_sub))
print(ops_sub["operacoes"].describe(percentiles=[0.05, 0.1, 0.25, 0.5, 0.75, 0.9]))

# [NOVO] As 10 sub-regiões com menos operações e as 10 com mais
display(ops_sub.sort_values("operacoes").head(10))
display(ops_sub.sort_values("operacoes").tail(10))
```

```python
# [NOVO - GJ00-54] Compara valores de K para escolher e explicar o K usando os dados

# [NOVO] Valores de K que vamos testar
ks_teste = [250_000, 500_000, 1_000_000, 1_500_000, 3_000_000]

linhas = []
for k in ks_teste:
    # [NOVO] Peso = operações / (operações + K), um peso para cada sub-região
    peso = ops_sub["operacoes"] / (ops_sub["operacoes"] + k)

    # [NOVO] Guarda o resumo deste K
    linhas.append({
        "K": k,
        "peso_menor": round(peso.min(), 2),                       # peso da sub-região menor
        "peso_mediana": round(peso.median(), 2),                  # peso da sub-região do meio
        "peso_maior": round(peso.max(), 2),                       # peso da sub-região maior
        "qtd_peso_acima_0.9": int((peso > 0.9).sum()),            # aceite do PO: quantas passam de 0,9
        "pct_peso_acima_0.9": round((peso > 0.9).mean() * 100, 1) # e em % do total
    })

# [NOVO] Mostra a comparação em uma tabela
display(pd.DataFrame(linhas))

# [NOVO] Confere o aceite do PO com o K sugerido: o peso das sub-regiões grandes (25% com mais operações)
K_TESTE = 1_000_000
limite_grandes = ops_sub["operacoes"].quantile(0.75)
grandes = ops_sub[ops_sub["operacoes"] >= limite_grandes]
peso_grandes = grandes["operacoes"] / (grandes["operacoes"] + K_TESTE)
print(f"K={K_TESTE:,}: menor peso entre as {len(grandes)} sub-regiões grandes = {peso_grandes.min():.3f}")
```

```python
# [NOVO - GJ00-54] Ajuste final: valor ajustado = peso x sub-região + (1 - peso) x estado

K = 1_000_000   # [NOVO] K escolhido (ver justificativa no final deste arquivo)

# [NOVO] Últimos 3 meses e os mesmos 3 meses do ano anterior (202607 -> 202507)
meses_atual = ultimos_3_meses
meses_ant = [str(int(m) - 100) for m in meses_atual]

# [NOVO] Coluna auxiliar: carteira ativa só do risco D-H (base da fragilidadeRenda)
df_scr["ativa_dh"] = np.where(df_scr["classe"] == "D-H", df_scr["carteira_ativa"], 0.0)

# [NOVO] Soma carteiras dos meses escolhidos. chaves = ["uf","sub_regiao"] (sub-região) ou ["uf"] (estado)
def somar(meses, chaves):
    d = df_scr[df_scr["data_base"].isin(meses)]
    return d.groupby(chaves)[["carteira_ativa", "carteira_vencida", "ativa_dh"]].sum().reset_index()

# [NOVO] Calcula os 3 indicadores do SCR; sem dado = NaN (nunca 0)
def indicadores(chaves):
    atual = somar(meses_atual, chaves)
    ant = somar(meses_ant, chaves)[chaves + ["carteira_vencida"]].rename(columns={"carteira_vencida": "vencida_ant"})
    df = atual.merge(ant, on=chaves, how="left")   # sem histórico do ano anterior -> vencida_ant = NaN

    # inadimplência = vencida / ativa
    df["inadimplenciaReal"] = np.where(df["carteira_ativa"] > 0, df["carteira_vencida"] / df["carteira_ativa"], np.nan)
    # fragilidade = ativa de risco D-H / ativa (se não há D-H no período, fica sem dado)
    df["fragilidadeRenda"] = np.where(df["ativa_dh"] > 0, df["ativa_dh"] / df["carteira_ativa"], np.nan)
    # crescimento da vencida contra o mesmo período do ano anterior
    df["agingDivida"] = np.where(df["vencida_ant"] > 0, df["carteira_vencida"] / df["vencida_ant"] - 1, np.nan)
    return df[chaves + ["inadimplenciaReal", "fragilidadeRenda", "agingDivida"]]

sub = indicadores(["uf", "sub_regiao"])
est = indicadores(["uf"])

# [NOVO] Junta: sub-região + valor do estado + operações (ops_sub vem da célula de contagem de operações)
df_aj = sub.merge(est, on="uf", suffixes=("", "_uf")).merge(ops_sub, on=["uf", "sub_regiao"])

# [NOVO] Peso = operações / (operações + K). Esse é o peso que o site vai usar para mostrar o aviso
df_aj["peso"] = df_aj["operacoes"] / (df_aj["operacoes"] + K)

# [NOVO] Valor ajustado. Se a sub-região ou o estado não tem dado, o resultado fica NaN (sem dado)
for ind in ["inadimplenciaReal", "fragilidadeRenda", "agingDivida"]:
    df_aj[ind + "_ajustado"] = df_aj["peso"] * df_aj[ind] + (1 - df_aj["peso"]) * df_aj[ind + "_uf"]

# [NOVO] Salva com o peso para o site
df_aj.to_parquet("/content/dados/subregioes_ajustadas.parquet", index=False)

# [NOVO] Conferência: 3 sub-regiões com menos e 3 com mais operações
cols = ["uf", "sub_regiao", "operacoes", "peso", "inadimplenciaReal", "inadimplenciaReal_uf", "inadimplenciaReal_ajustado"]
display(df_aj.sort_values("operacoes").head(3)[cols])
display(df_aj.sort_values("operacoes").tail(3)[cols])

# [NOVO] Aceite do PO: peso > 0,9 nas sub-regiões grandes (25% com mais operações)
grandes = df_aj[df_aj["operacoes"] >= df_aj["operacoes"].quantile(0.75)]
print(f"Menor peso entre as grandes: {grandes['peso'].min():.3f} -> {'OK' if grandes['peso'].min() > 0.9 else 'FALHOU'}")
print("Sem dado (NaN):", df_aj[[c for c in df_aj.columns if c.endswith('_ajustado')]].isna().sum().to_dict())
```

# **Visualização Gráfica – Eixo I: Risco de Crédito**

### Critérios de Aceite Atendidos:
- **Gráficos dos 4 indicadores:** Inadimplência Real, Fragilidade de Renda, Aging da Dívida e Vulnerabilidade Social.

- **Formatação completa:** Cada gráfico contém título, eixos nomeados (X e Y) e legenda explicativa.

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# (coluna, nome, cor da barra, cor da borda)
INDICADORES_EIXO_I = [
    ("inadimplenciaReal", "Inadimplência Real", "#0284c7", "#0369a1"),
    ("fragilidadeRenda", "Fragilidade de Renda", "#16a34a", "#15803d"),
    ("agingDivida", "Aging da Dívida", "#8b5cf6", "#6d28d9"),
    ("vulnerabilidadeSocial", "Vulnerabilidade Social", "#f59e0b", "#d97706"),
]


def visualizar_indicadores_eixo_i(df_dados, periodo="Todos", regiao="Todas"):
    """
    Gera os 4 gráficos do Eixo I - Risco de Crédito (por estado) com filtros de período e região.
    1. Gráficos de inadimplência real, fragilidade de renda, aging da dívida e vulnerabilidade social
    2. Filtros por período ("Todos", "2024" ou "202412") e região (ou UF)
    3. Título, eixos nomeados e legenda
    Valor 0 no df_eixo_i = "sem dado" (a escala válida é de 1 a 5): fica fora da média do período.
    """
    cols = [c for c, _, _, _ in INDICADORES_EIXO_I]
    df_f = df_dados.copy()
    for c in ["ano_mes", "regiao", "uf"]:
        df_f[c] = df_f[c].astype(str)

    # 1. Filtro por Período
    if periodo and str(periodo).lower() not in ["todos", "todas", "all"]:
        periodo_str = str(periodo)
        if len(periodo_str) == 4:
            df_f = df_f[df_f["ano_mes"].str.startswith(periodo_str)]
        else:
            df_f = df_f[df_f["ano_mes"] == periodo_str]

    # 2. Filtro por Região (ou UF)
    if regiao and str(regiao).lower() not in ["todas", "todos", "brasil", "todas as regiões", "all"]:
        regiao_str = str(regiao).strip()
        if regiao_str in df_f["regiao"].values:
            df_f = df_f[df_f["regiao"] == regiao_str]
        elif regiao_str in df_f["uf"].values:
            df_f = df_f[df_f["uf"] == regiao_str]
        else:
            df_f = df_f.iloc[0:0]

    if df_f.empty:
        print(f"Nenhum registro para Período='{periodo}' e Região='{regiao}'.")
        return None

    # Mês sem dado (valor 0) fica fora da média; a legenda mostra quantos meses entraram
    df_f[cols] = df_f[cols].where(df_f[cols] > 0)
    meses_total = df_f["ano_mes"].nunique()
    meses_com_dado = {c: df_f.loc[df_f[c].notna(), "ano_mes"].nunique() for c in cols}

    df_agg = df_f.groupby(["uf", "regiao"], as_index=False)[cols].mean().sort_values("uf")

    fig, axes = plt.subplots(2, 2, figsize=(18, 11))
    fig.patch.set_facecolor("#f8fafc")

    plt.suptitle(
        f"Eixo I – Indicadores de Risco de Crédito\n[Período: {periodo} | Região: {regiao}]",
        fontsize=16, fontweight="bold", color="#0f172a", y=0.98
    )

    for i, (ax, (col, nome, cor, borda)) in enumerate(zip(axes.flat, INDICADORES_EIXO_I), start=1):
        ax.set_facecolor("#ffffff")
        ax.set_title(f"{i}. {nome} por Estado", pad=10, fontweight="bold")
        ax.set_xlabel("Unidade Federativa (UF)", fontweight="bold")
        ax.set_ylabel(f"Índice de {nome} (Escala 1 a 5)", fontweight="bold")
        ax.grid(True, linestyle=":", alpha=0.6)

        dados = df_agg[["uf", col]].dropna()
        if dados.empty:
            ax.text(0.5, 0.5, "Sem dados neste período", ha="center", va="center",
                    transform=ax.transAxes, fontsize=12, color="gray")
            ax.set_xticks([])
            ax.set_yticks([])
            continue

        meses_txt = f" – {meses_com_dado[col]}/{meses_total} meses" if meses_com_dado[col] < meses_total else ""
        barras = ax.bar(dados["uf"], dados[col], color=cor, alpha=0.85, edgecolor=borda,
                        label=f"{nome} (Normalizado 1-5){meses_txt}")
        media = dados[col].mean()
        ax.axhline(media, color="#dc2626", linestyle="--", linewidth=1.5, label=f"Média ({media:.2f})")
        ax.set_ylim(0, 5.5)
        if len(dados) < 3:  # poucas barras (ex.: uma UF só) não devem ocupar o gráfico inteiro
            ax.set_xlim(-1.5, len(dados) + 0.5)
        ax.legend(loc="upper right", frameon=True)
        for b in barras:
            h = b.get_height()
            ax.annotate(f"{h:.2f}", (b.get_x() + b.get_width() / 2, h), xytext=(0, 3), textcoords="offset points",
                        ha="center", va="bottom", fontsize=8, fontweight="bold")

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()


# Abre os filtros (dropdowns) no topo, como no Eixo II
if "df_eixo_i" in globals():
    meses = sorted(df_eixo_i["ano_mes"].astype(str).unique(), reverse=True)
    anos = sorted({m[:4] for m in meses}, reverse=True)
    interact(
        lambda periodo, regiao: visualizar_indicadores_eixo_i(df_eixo_i, periodo, regiao),
        periodo=widgets.Dropdown(options=["Todos"] + anos + meses, value="Todos", description="Período:"),
        regiao=widgets.Dropdown(options=["Todas"] + sorted(df_eixo_i["regiao"].astype(str).unique()),
                                value="Todas", description="Região:"),
    )
else:
    print("Rode a EXECUÇÃO PRINCIPAL (gera o df_eixo_i) e depois esta célula de novo.")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def visualizar_indicadores_eixo_ii(df_dados, periodo="Todos", regiao="Todas", salvar_caminho=None):
    """
    Gera os 4 gráficos do Eixo II - Inclusão e Demografia com filtros de período e região.
    Atende integralmente aos critérios de aceite:
    1. Gráficos de maturidade Pix, crescimento populacional, total de habitantes e bônus demográfico
    2. Filtros por período e região
    3. Título, eixos nomeados e legenda
    """
    df_f = df_dados.copy()

    # 1. Filtro por Período
    if periodo and str(periodo).lower() not in ["todos", "todas", "all"]:
        periodo_str = str(periodo)
        if len(periodo_str) == 4:
            df_f = df_f[df_f["ano_mes"].astype(str).str.startswith(periodo_str)]
        else:
            df_f = df_f[df_f["ano_mes"].astype(str) == periodo_str]

    # 2. Filtro por Região
    if regiao and str(regiao).lower() not in ["todas", "todos", "brasil", "todas as regiões", "all"]:
        regiao_str = str(regiao).strip()
        if regiao_str in df_f["regiao"].values:
            df_f = df_f[df_f["regiao"] == regiao_str]
        elif regiao_str in df_f["uf"].values:
            df_f = df_f[df_f["uf"] == regiao_str]

    if df_f.empty:
        print(f"Nenhum registro para Período='{periodo}' e Região='{regiao}'.")
        return None

    df_agg = df_f.groupby(["uf", "regiao"], as_index=False).agg({
        "maturidadePix": "mean",
        "crescimentoPopulacional": "mean",
        "totalHabitantes": "mean",
        "bonusDemografico": "mean"
    }).sort_values("uf")

    ufs = df_agg["uf"].tolist()

    fig, axes = plt.subplots(2, 2, figsize=(18, 11))
    fig.patch.set_facecolor("#f8fafc")

    plt.suptitle(
        f"Eixo II – Indicadores de Inclusão e Expansão Demográfica\n[Período: {periodo} | Região: {regiao}]",
        fontsize=16, fontweight="bold", color="#0f172a", y=0.98
    )

    # 1. Maturidade Pix
    ax1 = axes[0, 0]
    ax1.set_facecolor("#ffffff")
    b1 = ax1.bar(ufs, df_agg["maturidadePix"], color="#0284c7", alpha=0.85, edgecolor="#0369a1", label="Maturidade Pix (Normalizado 1-5)")
    ax1.axhline(df_agg["maturidadePix"].mean(), color="#dc2626", linestyle="--", linewidth=1.5, label=f"Média ({df_agg['maturidadePix'].mean():.2f})")
    ax1.set_title("1. Maturidade Pix por Estado", pad=10, fontweight="bold")
    ax1.set_xlabel("Unidade Federativa (UF)", fontweight="bold")
    ax1.set_ylabel("Índice de Maturidade Pix (Escala 1 a 5)", fontweight="bold")
    ax1.set_ylim(0, 5.5)
    ax1.legend(loc="upper right", frameon=True)
    ax1.grid(True, linestyle=":", alpha=0.6)
    for b in b1:
        h = b.get_height()
        ax1.annotate(f"{h:.2f}", (b.get_x() + b.get_width() / 2, h), xytext=(0, 3), textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")

    # 2. Crescimento Populacional
    ax2 = axes[0, 1]
    ax2.set_facecolor("#ffffff")
    b2 = ax2.bar(ufs, df_agg["crescimentoPopulacional"], color="#16a34a", alpha=0.85, edgecolor="#15803d", label="Crescimento Populacional (Normalizado 1-5)")
    ax2.axhline(df_agg["crescimentoPopulacional"].mean(), color="#dc2626", linestyle="--", linewidth=1.5, label=f"Média ({df_agg['crescimentoPopulacional'].mean():.2f})")
    ax2.set_title("2. Crescimento Populacional por Estado (Censo IBGE)", pad=10, fontweight="bold")
    ax2.set_xlabel("Unidade Federativa (UF)", fontweight="bold")
    ax2.set_ylabel("Índice de Crescimento Populacional (Escala 1 a 5)", fontweight="bold")
    ax2.set_ylim(0, 5.5)
    ax2.legend(loc="upper right", frameon=True)
    ax2.grid(True, linestyle=":", alpha=0.6)
    for b in b2:
        h = b.get_height()
        ax2.annotate(f"{h:.2f}", (b.get_x() + b.get_width() / 2, h), xytext=(0, 3), textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")

    # 3. Total de Habitantes
    ax3 = axes[1, 0]
    ax3.set_facecolor("#ffffff")
    b3 = ax3.bar(ufs, df_agg["totalHabitantes"], color="#8b5cf6", alpha=0.85, edgecolor="#6d28d9", label="Total de Habitantes (Normalizado 1-5)")
    ax3.axhline(df_agg["totalHabitantes"].mean(), color="#dc2626", linestyle="--", linewidth=1.5, label=f"Média ({df_agg['totalHabitantes'].mean():.2f})")
    ax3.set_title("3. Total de Habitantes por Estado (População Residente)", pad=10, fontweight="bold")
    ax3.set_xlabel("Unidade Federativa (UF)", fontweight="bold")
    ax3.set_ylabel("Índice de Total de Habitantes (Escala 1 a 5)", fontweight="bold")
    ax3.set_ylim(0, 5.5)
    ax3.legend(loc="upper right", frameon=True)
    ax3.grid(True, linestyle=":", alpha=0.6)
    for b in b3:
        h = b.get_height()
        ax3.annotate(f"{h:.2f}", (b.get_x() + b.get_width() / 2, h), xytext=(0, 3), textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")

    # 4. Bônus Demográfico
    ax4 = axes[1, 1]
    ax4.set_facecolor("#ffffff")
    b4 = ax4.bar(ufs, df_agg["bonusDemografico"], color="#f59e0b", alpha=0.85, edgecolor="#d97706", label="Bônus Demográfico (Normalizado 1-5)")
    ax4.axhline(df_agg["bonusDemografico"].mean(), color="#dc2626", linestyle="--", linewidth=1.5, label=f"Média ({df_agg['bonusDemografico'].mean():.2f})")
    ax4.set_title("4. Bônus Demográfico por Estado (Janela Etária)", pad=10, fontweight="bold")
    ax4.set_xlabel("Unidade Federativa (UF)", fontweight="bold")
    ax4.set_ylabel("Índice de Bônus Demográfico (Escala 1 a 5)", fontweight="bold")
    ax4.set_ylim(0, 5.5)
    ax4.legend(loc="upper right", frameon=True)
    ax4.grid(True, linestyle=":", alpha=0.6)
    for b in b4:
        h = b.get_height()
        ax4.annotate(f"{h:.2f}", (b.get_x() + b.get_width() / 2, h), xytext=(0, 3), textcoords="offset points", ha="center", va="bottom", fontsize=8, fontweight="bold")

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    if salvar_caminho:
        plt.savefig(salvar_caminho, dpi=200, bbox_inches="tight")
    plt.show()
```
